```python
# Classes and Function (3 marks)

# Step 1: Create a class BankAccount.
# Add class attribute bankname="Canada Bank"
# Add 2 instance attributes: account_holder (string) & balance (float, default value 0.0)

class BankAccount:
    bankname = "Canada Bank"  #class attribute

    def __init__(self, account_holder, balance=0.0):
        self.account_holder = account_holder
        self.balance = balance

# Step 2: Add methods
# 1. deposit(amount) -> Add amount to the balance
# 2. withdraw(amount) -> Subtract the amount if balance is sufficient, else print an error message
# 3. get_balance() -> Print Account holder's name and current balance
# 4. get_bankname() -> Print the class attribute (staticmethod)
# Note: Use f-strings to display the amount and updated balance whenever you call deposit() or withdraw()

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount} CAD. Updated balance: {self.balance} CAD")

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"Withdrew {amount} CAD. Updated balance: {self.balance} CAD")
        else:
            print(f"Error: Insufficient funds. Current balance: {self.balance} CAD")

    def get_balance(self):
        print(f"Account holder: {self.account_holder}, Balance: {self.balance} CAD")

    @staticmethod
    def get_bankname():
        print(f"Bank Name: {BankAccount.bankname}")


# Step 3: Test
# 1. Print the Bankname

BankAccount.get_bankname()

# 2. Create an object of BankAccount with your name and an initial balance of 1000

account1 = BankAccount("Khushi Mishra", 1000)

# 3. Deposit 500

account1.deposit(500)

# 4. Withdraw 200

account1.withdraw(200)

# 5. Print the final balance

account1.get_balance()
```

```python
# Merge the studentHomerooms with the merged dataset & name it df_studentInfo. Make sure that the key you are joining on are the same typecast for both datasets.
# Also ensure that there are no redundant columns, and finally display the first few rows.
# Hint: try joining on more than one column to avoid having to drop redundant columns, ie. on=['studentID', '<ColumnA>', '<ColumnB>'] ...
# (5 marks)

#Checking the column in both datasets
print(df_studentGrades.columns)
print(df_homeroom.columns)

# Ensureing that studentID is the same type in both datasets
df_studentGrades['studentID'] = df_studentGrades['studentID'].astype(str)
df_homeroom['studentID'] = df_homeroom['studentID'].astype(str)

# Merging the datasets on studentID
df_studentInfo = pd.merge(
    df_studentGrades,
    df_homeroom,
    left_on=['studentID', 'firstName_English', 'lastName_English'],
    right_on=['studentID', 'firstName', 'lastName'],
    how='inner'
)
# To avoid dropping unnecessary columns, trying to join on multiple columns.
# Only keep essential columns (optional renaming)
df_studentInfo = df_studentInfo[['firstName_English', 'lastName_English', 'studentID',
                                 'grade_English', 'grade_Math', 'GPA', 'homeRoom']]

# Display first few rows
print(df_studentInfo.head())

```

```python
# The code below calculates the mean of each homeroom class.
df_grouped = df_studentInfo.groupby('homeRoom')['GPA'].mean()

# Printing the mean GPA per homeroom
print(df_grouped)

# Which homeroom has a higher GPA?
# (1 mark)
# Identify the homeroom with the highest mean GPA
highest_gpa_homeroom = df_grouped.idxmax()
print(f"Homeroom with the highest GPA: {highest_gpa_homeroom}")
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''

  # ----- Step 1: Build design matrix A -----
    # A has columns [1, a, a^2, ..., a^deg]
    A = np.vander(a, N=deg+1, increasing=True) #function in NumPy creates a Vandermonde matrix

    # ----- Step 2: Apply normal equation (Weighted) -----
    # x = (A^T P A)^(-1) A^T P l
    AtPA = A.T @ P @ A
    AtPA_inv = np.linalg.pinv(AtPA)
    x = AtPA_inv @ (A.T @ P @ l)

    # ----- Step 3: Compute residuals -----
    # v = l - A x
    v = l - A @ x

    # ----- Step 4: A-posteriori variance factor -----
    # apv = v^T P v / (n - u)
    n = len(l)          # number of observations
    u = deg + 1         # number of unknown parameters
    apv = (v.T @ P @ v) / (n - u)

    # ----- Step 5: Covariance of estimated parameters -----
    # Cov(x) = apv * (A^T P A)^(-1)
    cov_x = apv * AtPA_inv

    return (x, cov_x, apv, v)
```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)
deg_list = [2, 4, 6]

# store results
results_uw = {}  # uw = unweighted

for deg in deg_list:
    x, cov_x, apv, v = fit_polynomial_curve(a, l, P, deg)
    results_uw[deg] = (x, cov_x, apv, v)
    print(f"Degree {deg} fit complete. a-posteriori variance factor = {apv}")
```

```python
# step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 marks)

# degree 2 polynomial (quadratic)
quadratic_uw = np.polyval(results_uw[2][0][::-1], a)

# degree 4 polynomial (quartic)
quartic_uw = np.polyval(results_uw[4][0][::-1], a)

# degree 6 polynomial (hexic)
hexic_uw = np.polyval(results_uw[6][0][::-1], a)
```

```python
# step 10
# compute and print the norm between the predicted observations (output from each polynomial function) and the true observations
# (3 marks)

x = df_data['l'].values
y_true = df_data['l_true'].values

degrees = [1, 2, 3, 4, 5]
norms = []

for d in degrees:
    # Create polynomial features manually
    X_poly = np.vstack([x**i for i in range(d + 1)]).T

    # Solve for coefficients using least squares
    coeffs = np.linalg.lstsq(X_poly, y_true, rcond=None)[0]

    # Predict
    y_pred = X_poly @ coeffs

    # Compute L2 norm
    norm = np.linalg.norm(y_pred - y_true)
    norms.append(norm)
    print(f"Degree {d}: L2 norm = {norm:.2f}")
```

```python
# step 11
# comment on the results of a posterier variance factor and norm, what pattern do you observe
# (2 marks)
# ...

plt.figure(figsize=(8, 5))
plt.plot(degrees, norms, marker='o', linestyle='-', color='teal')
plt.title("L2 Norm of Prediction Error vs Polynomial Degree")
plt.xlabel("Polynomial Degree")
plt.ylabel("L2 Norm")
plt.grid(True)
plt.show()

# Posterior Variance Factor (l_var) reflects the uncertainty or confidence in each observation. Higher values suggest more reliable data points.
# Norm Trend: As polynomial degree increases, the norm decreases, indicating better fit. However, the improvement plateaus beyond degree 3, suggesting diminishing returns from added complexity.
# Predictions tend to align more closely with l_true when l_var is high, implying that observations with higher certainty contribute more effectively to model accuracy.
# The norm reduction is not linear—this suggests that while increasing model complexity helps, the influence of variance-weighted observations is crucial for meaningful improvements.
```

```python
# step 12
# now, let's define the P matrix with the proper weights of observation
# use the variance of observation l_var, define P matrix
# (1 mark)
w = 1 / df_data['l_var'].values  # weights = 1 / variance
P = np.diag(w)              # diagonal weight matrix
```

```python
# step 13
# repeat the steps 6-10, using the proper P matrix
# (7 marks)

x = df_data['l'].values
y_true = df_data['l_true'].values
degrees = [1, 2, 3, 4, 5]
weighted_norms = []

for d in degrees:
    # Create polynomial features
    X_poly = np.vstack([x**i for i in range(d + 1)]).T

    # Apply weighted least squares: β = (Xᵀ P X)⁻¹ Xᵀ P y
    XT_P = X_poly.T * w  # equivalent to Xᵀ @ P
    beta = np.linalg.inv(XT_P @ X_poly) @ (XT_P @ y_true)

    # Predict
    y_pred = X_poly @ beta

    # Compute weighted L2 norm: sqrt((y_pred - y_true)ᵀ P (y_pred - y_true))
    residual = y_pred - y_true
    norm = np.sqrt(residual.T @ P @ residual)
    weighted_norms.append(norm)
    print(f"Degree {d}: Weighted L2 norm = {norm:.2f}")
```

```python
# step 14
# Compare the results of a posterier variance factor and norm from unweighted LSE and weighted LSE, what phenomena do you observe? Explain this phenomena.
# (2 marks)

# Plot comparison
plt.figure(figsize=(8, 5))
plt.plot(degrees, norms, 'o--', label='Unweighted Norm', color='gray')
plt.plot(degrees, weighted_norms, 'o-', label='Weighted Norm (using P)', color='blue')
plt.title("Comparison of L2 Norms (Unweighted vs Weighted)")
plt.xlabel("Polynomial Degree")
plt.ylabel("L2 Norm")
plt.legend()
plt.grid(True)
plt.show()

# The weighted LSE consistently yields lower or more stable norms compared to unweighted LSE, especially for higher-degree polynomials.
# This shows that incorporating variance improves model robustness, reducing the influence of noisy observations.
# The norm reduction is more meaningful in weighted LSE because it reflects a better fit to the most trustworthy data.
```
