**Pseudorange** is the *pseudo* distance between a satellite and a satellite signal receiver on the ground. In simple words, it contains the geometric range between the satellite and the receiver, with additional associated errors.

The simplified version of the Pseudorange measurement model is:
$$P=\rho-cdT+ϵ_{sum}$$ and
$$\rho=\sqrt{(x_s - x_r)^2 + (y_s - y_r)^2 + (z_s - z_r)^2}$$
where
$$\rho=geometric\ range$$
$${x_s, y_s, z_s, x_r,y_r,z_r} = 3D\ coordinates\ of\ satellite\ and\ receiver$$
$$cdT = receiver\ clock\ offset$$
$$ϵ_{sum}=general\ errors\ term\ (ϵ_{sum}=\Delta \rho + \Delta_{\text{ion}} + \Delta_{\text{trop}} + \epsilon_P + cdt)$$

\
Note: the 3D **positions of satellites** are given in this assignment.

### **Unknown States:** Position of receiver & receiver clock offset

We use the psuedorange measurements to estimate the receiver's **3-dimensional position** and the **receiver clock offset:** $$X={x_r,y_r,z_r,cdT}$$.

### Step 4: define a function named design_matrix which computes the design matrix A of this LSE problem (3 mark)

the input of this funciton should be: estimated unknown states array, satellites positions array and geometric distance array

the function should return an two-dimensional array which represents the design matrix A

#### Calculation Formula for the Design Matrix \( A \)

Let the position vectors of the satellite and receiver be:

- **r** = (r_x, r_y, r_z) for the receiver
- **s** = (s_x, s_y, s_z) for the satellite

The Euclidean distance between the satellite and receiver is given by:

$$
\rho = \sqrt{(s_x - r_x)^2 + (s_y - r_y)^2 + (s_z - r_z)^2}
$$

#### Partial Derivatives of the Distance with Respect to \(x\), \(y\), and \(z\):

To construct the design matrix, we compute the partial derivatives of the distance function with respect to the receiver's \(x\), \(y\), and \(z\) coordinates. Here's the detailed process for one of them, and the others follow the same procedure.

##### Partial derivative with respect to \(r_x\):

Taking the derivative with respect to \(r_x\), we use the chain rule:

$$
\frac{\partial \rho}{\partial r_x} = \frac{\partial}{\partial r_x} \left( \sqrt{(s_x - r_x)^2 + (s_y - r_y)^2 + (s_z - r_z)^2} \right)
$$

First, apply the derivative of the square root:

$$
\frac{\partial \rho}{\partial r_x} = \frac{1}{2 \sqrt{(s_x - r_x)^2 + (s_y - r_y)^2 + (s_z - r_z)^2}} \cdot 2 (r_x - s_x)
$$

Simplifying:

$$
\frac{\partial \rho}{\partial r_x} = \frac{r_x - s_x}{\rho}
$$

##### Partial derivatives for \(r_y\) and \(r_z\):

The partial derivatives with respect to \(r_y\) and \(r_z\) follow the same procedure:

$$
\frac{\partial \rho}{\partial r_y} = \frac{r_y - s_y}{\rho}
$$

$$
\frac{\partial \rho}{\partial r_z} = \frac{r_z - s_z}{\rho}
$$

#### The Design Matrix \(A\):

Thus, the design matrix \(A\) can be expressed as:

$$
A = \left[
\begin{array}{ccc|c}
\frac{s_x - r_x}{\text{distance}} & \frac{s_y - r_y}{\text{distance}} & \frac{s_z - r_z}{\text{distance}} & -1
\end{array}
\right]
$$

Where:
- The first three columns represent the partial derivatives of the distance with respect to x, y, and z coordinates
- The last column of -1 corresponds to the constant term in the least squares equation

This matrix is used for least squares estimation in positioning or navigation applications.


```python
def design_matrix(receiver_pos, satellite_pos, distance):

    # Build the design matrix A (N x 4):
    #  columns = [dρ/dx, dρ/dy, dρ/dz, dρ/dbias] = [(sx-rx)/ρ, (sy-ry)/ρ, (sz-rz)/ρ, -1]
    # Accepts satellite_pos with columns either [x,y,z] or [Satellite_ID,x,y,z].
    # 'distance' must be a length-N vector of ρ values (geometric ranges).

    rec = np.asarray(receiver_pos, float).reshape(3)
    sat = np.asarray(satellite_pos, float)
    if sat.ndim != 2:
        raise ValueError(f"satellite_pos must be 2D, got shape {sat.shape}")
    if sat.shape[1] == 4:           # drop ID if present
        sat = sat[:, 1:4]
    if sat.shape[1] != 3:
        raise ValueError(f"satellite_pos must have 3 columns (x,y,z). Got shape {sat.shape}")

    rho = np.asarray(distance, float).reshape(-1)
    if rho.shape[0] != sat.shape[0]:
        raise ValueError(
            f"distance length ({rho.shape[0]}) must match number of satellites ({sat.shape[0]})."
        )

    sx, sy, sz = sat[:, 0], sat[:, 1], sat[:, 2]
    rx, ry, rz = rec
    eps = 1e-12
    #rho = np.maximum(rho, eps)
    rho = np.maximum(np.asarray(distance, float), 1e-12)

    A = np.column_stack([
        (sx - rx) / rho,
        (sy - ry) / rho,
        (sz - rz) / rho,
        -np.ones_like(rho)
    ])
    return A
```

### Step 5: start the iteration for the LSE (4 mark)

define the convergence criteria for the 'while' loop

the convergence criteria is that the corrections delta are less than 0.008 (i.e., <0.1 m), for receiver's position x,y,z part

```python
GroundTruth = np.array([-1633489.37967725, -3651627.18250363, 4952481.61954918])
n = pseudorange.shape[0]
m = 4
sigma2 = 1.2
P = (1.0/sigma2) * np.eye(n)

# rough Earth-surface guess near the sat centroid:
x_hat = np.array([0.0, 0.0, 6371000.0, 0.0])     # meters

delta = np.full(4, 10.0)
i = 0
max_iters = 100

arr_w, arr_delta, arr_err_pos, arr_i = [], [], [], []

# --- 3) Step 5: LSE iteration until max(|delta_xyz|) < 0.008 m ---
while (np.max(np.abs(delta[:3])) >= 0.008) and (i < max_iters):
    # distances and design matrix
    distance = euclidean_norm(x_hat[:3], sat_pos)            # (N,)
    A        = design_matrix(x_hat[:3], sat_pos, distance)   # (N,4)

    # misclosure w = observed - computed; computed = range + b
    computed = distance + x_hat[3]
    w        = pseudorange - computed                        # (N,)

    # normal equations
    N = A.T @ P @ A
    u = A.T @ P @ w
    try:
        delta = np.linalg.solve(N, u)
    except np.linalg.LinAlgError:
        delta, *_ = np.linalg.lstsq(N, u, rcond=None)

    # (optional) simple trust-region clamp to avoid wild steps
    step_norm = np.linalg.norm(delta[:3])
    if step_norm > 5e4:          # 50 km cap per iteration (tune if you like)
        delta[:3] *= 5e4 / step_norm

    # state update
    x_hat = x_hat + delta

    # position error vs ground truth
    err_pos = np.linalg.norm(x_hat[:3] - GroundTruth)

    # iterate & store histories
    i += 1
    arr_w.append(float(np.linalg.norm(w)))
    arr_delta.append(float(np.linalg.norm(delta[:3])))
    arr_err_pos.append(float(err_pos))
    arr_i.append(i)

# --- 4) After convergence: residuals, APV, covariance ---
# v   = pseudorange - (euclidean_norm(x_hat[:3], sat_pos) + x_hat[3])
# apv = float(v.T @ P @ v) / (n - m)
# C_x_hat = apv * np.linalg.inv((design_matrix(x_hat[:3], sat_pos, euclidean_norm(x_hat[:3], sat_pos))).T @ P @
                              #design_matrix(x_hat[:3], sat_pos, euclidean_norm(x_hat[:3], sat_pos)))

v   = pseudorange - (euclidean_norm(x_hat[:3], sat_pos) + x_hat[3])
dof = len(pseudorange) - 4
if dof <= 0:
    raise ValueError("Need at least 5 sats for reliable covariance.")
apv = float(v.T @ P @ v) / dof
C_x_hat   = apv * np.linalg.inv(A.T @ P @ A)  # A from final iteration


print("Converged in", i, "iterations; max |delta_xyz| =", np.max(np.abs(delta[:3])))
print("\nx_hat [X,Y,Z,b] (m):", np.round(x_hat, 4))
print("RMS residual (m):", float(np.sqrt(np.mean(v**2))))
print("APV (sigma0^2):", apv)
print("Final 3D error (m):", float(np.linalg.norm(x_hat[:3] - GroundTruth)))
```

### Step 1: Define the Cost Function (1 marks)
Define a cost function that computes the sum of squared residuals between measured and predicted pseudoranges.

The function's inputs are: the receiver's coordinates, clock offset, satellites' coordinates, observed pseudoranges.

The function returns a single scalar.

Call euclidean_norm function to calculate geometric distance.

```python
def cost_function(rec_pos, cdT, sat_pos, pseudorange):

   # Sum of squared residuals:  L = sum_i (rho_obs_i - (rho_geom_i + cdT))^2
   # rec_pos: (3,)
   # cdT: float (meters)
   # sat_pos: (N,3)
   # pseudorange: (N,)

    rho_geom = euclidean_norm(rec_pos, sat_pos)   # (N,)
    pred     = rho_geom + cdT                     # (N,)
    r        = pseudorange - pred                 # (N,)
    return float(np.dot(r, r))
```

### Step 2: Define the Gradient Computation Function (1 marks)

Define a gradient computation function that calculates the gradient of the cost function with respect to the estimates (x_r, y_r, z_r, cdT).


The function's inputs are: the receiver's coordinates, clock offset, satellites' coordinates, observed pseudoranges.


The function returns a 1D numpy array that contains 4 elements.


Call euclidean_norm function to calculate geometric distance.


**Recall elements of the design matrix A in the previous LSE function.**

```python
def gradient(rec_pos, cdT, sat_pos, pseudorange):

    # Gradient of L w.r.t [x, y, z, cdT].
    # Uses: d(rho)/d(rx) = (rx - sx)/rho, etc. and d(pred)/d(cdT)=1

    rho = euclidean_norm(rec_pos, sat_pos)                 # (N,)
    rho = np.maximum(rho, 1e-12)

    rx, ry, rz = rec_pos
    sx, sy, sz = sat_pos[:,0], sat_pos[:,1], sat_pos[:,2]

    # Jacobian of the prediction h = rho + cdT:
    H = np.column_stack([
        (rx - sx)/rho,    # d rho / d rx
        (ry - sy)/rho,    # d rho / d ry
        (rz - sz)/rho,    # d rho / d rz
        np.ones_like(rho) # d (cdT)/d cdT
    ])                                  # (N,4)

    r = pseudorange - (rho + cdT)       # residuals (N,)
    grad = -2.0 * (H.T @ r)             # (4,)
    return grad
```

### Step 3: Gradient Descent Implementation (4 marks)
Construct the implementation function of gradient descent

Replace '...' with actual implementation

The function's input are: observed pseudoranges, satellites' positions, learning_rate, maximum loop iterations, tolerance

The function will return a 1D numpy array that contains the optimized estimates (x_r, y_r, z_r, cdT)

The function set learning_rate, max_iters and tolerance as: 0.08, 10000, 8e-7 as default values

These default hyperparameters values can be overridden by passing your own values when calling the function


```python

def gradient_descent(pseudorange, sat_pos, learning_rate=0.08, max_iters=10000, tolerance=8e-7):
    global steps_count
    X = np.array([50.0, -25.0, 15.0, 500.0], dtype=float)  # init

    arr_cost.clear(); arr_grad.clear(); arr_err.clear()

    for iteration in range(max_iters):
        rec_pos, cdT = X[:3], X[3]

        current_cost = cost_function(rec_pos, cdT, sat_pos, pseudorange)
        g = gradient(rec_pos, cdT, sat_pos, pseudorange)

        # update
        step = learning_rate * g
        X_new = X - step

        # record
        arr_cost.append(current_cost)
        arr_grad.append(g)
        # if you have GroundTruth:
        # arr_err.append(X[:3] - GroundTruth)

        # convergence on step size
        if np.linalg.norm(step) < tolerance:
            steps_count = iteration
            X = X_new
            break

        X = X_new

    return X
```

```python

# Run optimization with default hyperparameters
optimized_params = gradient_descent(pseudorange, sat_pos)

# Extract optimized receiver coordinates and clock bias
opt_rec_pos = optimized_params[:3]
opt_cdT     = optimized_params[3]

# Final metrics
final_cost = cost_function(opt_rec_pos, opt_cdT, sat_pos, pseudorange)
final_grad = gradient(opt_rec_pos, opt_cdT, sat_pos, pseudorange)

print("Optimized parameters [x, y, z, cdT] (m):")
print(optimized_params)
print("\nSteps to convergence:", steps_count)
print("Final cost:", final_cost)
print("Final gradient:", final_grad)

# If you have GroundTruth defined:
# print("Final position error (m):", np.linalg.norm(opt_rec_pos - GroundTruth))
```

#### Fundamental Difference between MLP and CCN:

**Key Architectural Differences:**
- **MLP**: Input → Flatten → Dense layers → Output (treats pixels independently)
- **CNN**: Input → Reshape → Conv+Pool layers → Flatten → Dense layers → Output (learns spatial patterns first)

**The CNN Advantage:**
1. **Convolutional Layers**: Extract spatial features (edges, curves, shapes) that MLP cannot detect
2. **Pooling Layers**: Provide translation invariance and reduce computational load
3. **Feature Hierarchy**: Early layers detect simple patterns, deeper layers combine them into complex features
4. **Parameter Sharing**: Same filter applied across entire image, making CNN more efficient

**Why CNNs End with Dense Layers:**
- Convolutional layers act as **feature extractors** (learn "what" is in the image)
- Dense layers act as **classifiers** (decide "which digit" based on extracted features)
- This is a common pattern: **Feature Extraction (CNN) + Classification (MLP)**

**The Real Difference:**
- **MLP Dense layers**: Work with raw pixel values (784 individual numbers)
- **CNN Dense layers**: Work with high-level spatial features learned by convolutional layers
